```python

# Identify cancellation invoices
processed_df["IsCancellation"] = (
    processed_df["InvoiceNo"].astype(str).str.startswith("C")
)

# Calculate transaction value (not profit)
processed_df["SalesValue"] = (
    processed_df["Quantity"] * processed_df["UnitPrice"]
)

# Flag ordinary positive-sale records
processed_df["IsPositiveSale"] = (
    (processed_df["Quantity"] > 0)
    & (processed_df["UnitPrice"] > 0)
    & (~processed_df["IsCancellation"])
)

# Flag selected special item codes for review
processed_df["IsSpecialItem"] = (
    processed_df["StockCode"].isin(["AMAZONFEE", "M", "D"])
)

print("Processed rows:", len(processed_df))
print("Cancellation invoices:", processed_df["IsCancellation"].sum())
print("Positive-sale rows:", processed_df["IsPositiveSale"].sum())
print("Special-item rows:", processed_df["IsSpecialItem"].sum())

display(processed_df.head())
```

```python

# Rebuild from the untouched original dataset
processed_df = df.copy()

# Remove exact duplicates first
processed_df = processed_df.drop_duplicates().copy()

# Clean text columns without converting numeric StockCodes to missing values
for col in ["Description", "Country"]:
    processed_df[col] = processed_df[col].str.strip()

# Preserve all StockCode values; trim only string values
processed_df["StockCode"] = processed_df["StockCode"].map(
    lambda x: x.strip() if isinstance(x, str) else x
)

# Fill missing descriptions
processed_df["Description"] = (
    processed_df["Description"].fillna("Unknown Description")
)

# Create analysis columns
processed_df["IsCancellation"] = (
    processed_df["InvoiceNo"].astype(str).str.startswith("C")
)

processed_df["SalesValue"] = (
    processed_df["Quantity"] * processed_df["UnitPrice"]
)

processed_df["IsPositiveSale"] = (
    (processed_df["Quantity"] > 0)
    & (processed_df["UnitPrice"] > 0)
    & (~processed_df["IsCancellation"])
)

processed_df["IsSpecialItem"] = (
    processed_df["StockCode"].astype(str).isin(
        ["AMAZONFEE", "M", "D"]
    )
)

# Verify the correction
print("Processed dataset shape:", processed_df.shape)
print("Missing StockCode:", processed_df["StockCode"].isna().sum())
print("Missing Description:", processed_df["Description"].isna().sum())
print("Remaining exact duplicates:", processed_df.duplicated().sum())
print("Missing CustomerID:", processed_df["CustomerID"].isna().sum())
```

```python

# Keep ordinary positive-sale records for sales analysis
sales_df = processed_df[
    processed_df["IsPositiveSale"]
].copy()

# Exclude selected special codes from ordinary product analysis
sales_df = sales_df[
    ~sales_df["IsSpecialItem"]
].copy()

print("Rows in sales analysis dataset:", len(sales_df))
print("Total sales value:", round(sales_df["SalesValue"].sum(), 2))
print("Unique invoices:", sales_df["InvoiceNo"].nunique())
print("Unique products:", sales_df["StockCode"].nunique())
print("Unique countries:", sales_df["Country"].nunique())
print("Date range:",
      sales_df["InvoiceDate"].min(),
      "to",
      sales_df["InvoiceDate"].max())

display(sales_df.head())
```

```python

# Create a month column
sales_df["Month"] = sales_df["InvoiceDate"].dt.to_period("M").astype(str)

# Summarize sales by month
monthly_sales = (
    sales_df.groupby("Month")
    .agg(
        TotalSales=("SalesValue", "sum"),
        InvoiceCount=("InvoiceNo", "nunique"),
        UnitsSold=("Quantity", "sum")
    )
    .reset_index()
)

# Sort chronologically
monthly_sales = monthly_sales.sort_values("Month")

# Display monthly results
monthly_sales["TotalSales"] = monthly_sales["TotalSales"].round(2)

display(monthly_sales)

# Identify the highest-sales month
top_month = monthly_sales.loc[monthly_sales["TotalSales"].idxmax()]

print("\nHighest-sales month:", top_month["Month"])
print("Sales value:", round(top_month["TotalSales"], 2))
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))
plt.plot(
    monthly_sales["Month"],
    monthly_sales["TotalSales"],
    marker="o"
)
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales Value (£)")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python

# Rank products by total sales value
top_products = (
    sales_df.groupby(["StockCode", "Description"])
    .agg(
        TotalSales=("SalesValue", "sum"),
        UnitsSold=("Quantity", "sum"),
        TransactionLines=("InvoiceNo", "count")
    )
    .reset_index()
    .sort_values("TotalSales", ascending=False)
    .head(10)
)

top_products["TotalSales"] = top_products["TotalSales"].round(2)

display(top_products)
```

```python

top_by_units = (
    sales_df.groupby(["StockCode", "Description"])
    .agg(UnitsSold=("Quantity", "sum"))
    .reset_index()
    .sort_values("UnitsSold", ascending=False)
    .head(10)
)

display(top_by_units)
```

```python

country_sales = (
    sales_df.groupby("Country")
    .agg(
        TotalSales=("SalesValue", "sum"),
        InvoiceCount=("InvoiceNo", "nunique"),
        UnitsSold=("Quantity", "sum"),
        UniqueCustomers=("CustomerID", "nunique")
    )
    .reset_index()
    .sort_values("TotalSales", ascending=False)
)

country_sales["TotalSales"] = country_sales["TotalSales"].round(2)

display(country_sales.head(10))

print("\nHighest-sales country:")
print(country_sales.iloc[0]["Country"])

print("Sales value:")
print(round(country_sales.iloc[0]["TotalSales"], 2))
```

```python

import matplotlib.pyplot as plt

top_countries = country_sales.head(10).sort_values("TotalSales")

plt.figure(figsize=(10, 6))
plt.barh(top_countries["Country"], top_countries["TotalSales"])
plt.title("Top 10 Countries by Sales Value")
plt.xlabel("Sales Value (£)")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```

```python

# Investigate the unusually large quantity
unusual_transaction = sales_df[
    (sales_df["StockCode"].astype(str) == "23843")
    & (sales_df["Quantity"] >= 10000)
]

display(
    unusual_transaction[
        ["InvoiceNo", "StockCode", "Description", "Quantity",
         "InvoiceDate", "UnitPrice", "CustomerID",
         "Country", "SalesValue"]
    ]
)

print("Matching rows:", len(unusual_transaction))
print("Total quantity:", unusual_transaction["Quantity"].sum())
print("Total sales value:",
      round(unusual_transaction["SalesValue"].sum(), 2))
```

```python

# Show the 10 largest positive quantities
largest_transactions = (
    sales_df.nlargest(10, "Quantity")[
        ["InvoiceNo", "StockCode", "Description", "Quantity",
         "InvoiceDate", "UnitPrice", "Country", "SalesValue"]
    ]
)

display(largest_transactions)
```

```python

# Keep records with a known customer ID
customer_df = sales_df.dropna(subset=["CustomerID"]).copy()

# Summarize purchasing behavior by customer
customer_sales = (
    customer_df.groupby("CustomerID")
    .agg(
        TotalSales=("SalesValue", "sum"),
        InvoiceCount=("InvoiceNo", "nunique"),
        UnitsPurchased=("Quantity", "sum")
    )
    .reset_index()
    .sort_values("TotalSales", ascending=False)
)

customer_sales["TotalSales"] = customer_sales["TotalSales"].round(2)

print("Customers analyzed:", len(customer_sales))
print("\nTop 10 customers by sales value:")
display(customer_sales.head(10))

print("\nAverage sales value per customer:",
      round(customer_sales["TotalSales"].mean(), 2))
```

```python
# Identify cancellation records
cancellations = processed_df[
    processed_df["IsCancellation"]
].copy()

print("Total cancellation records:", len(cancellations))
print("Unique cancelled invoices:", cancellations["InvoiceNo"].nunique())

print(
    "Cancellation quantity:",
    cancellations["Quantity"].sum()
)

print(
    "Net recorded value of cancellation rows (£):",
    round(cancellations["SalesValue"].sum(), 2)
)

print("\nFirst 10 cancellation records:")
display(
    cancellations[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "UnitPrice", "InvoiceDate", "Country"]
    ].head(10)
)
```

```python
# Analyze cancelled products
cancellation_products = (
    cancellations.groupby(["StockCode", "Description"])
    .agg(
        CancellationLines=("InvoiceNo", "count"),
        TotalCancelledQuantity=("Quantity", "sum"),
        RecordedCancellationValue=("SalesValue", "sum")
    )
    .reset_index()
    .sort_values("CancellationLines", ascending=False)
    .head(10)
)

cancellation_products["RecordedCancellationValue"] = (
    cancellation_products["RecordedCancellationValue"].round(2)
)

print("Top 10 products by number of cancellation records:")
display(cancellation_products)
```

```python
# Monthly positive sales
monthly_positive_sales = (
    sales_df.assign(
        Month=sales_df["InvoiceDate"].dt.to_period("M").astype(str)
    )
    .groupby("Month")["SalesValue"]
    .sum()
    .rename("PositiveSales")
)

# Monthly cancellation values
monthly_cancellation_values = (
    cancellations.assign(
        Month=cancellations["InvoiceDate"].dt.to_period("M").astype(str)
    )
    .groupby("Month")["SalesValue"]
    .sum()
    .rename("CancellationValue")
)

# Combine the monthly summaries
monthly_comparison = pd.concat(
    [monthly_positive_sales, monthly_cancellation_values],
    axis=1
).fillna(0).reset_index()

monthly_comparison["PositiveSales"] = (
    monthly_comparison["PositiveSales"].round(2)
)
monthly_comparison["CancellationValue"] = (
    monthly_comparison["CancellationValue"].round(2)
)

monthly_comparison["CancellationRatePercent"] = (
    monthly_comparison["CancellationValue"].abs()
    / monthly_comparison["PositiveSales"] * 100
).round(2)

display(monthly_comparison)
```
